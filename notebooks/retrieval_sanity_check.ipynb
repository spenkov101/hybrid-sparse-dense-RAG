```python
from src.retrieval import (
    HybridRetriever,
    create_search_config,
    get_top_result,
)
```

```python
passages = [
    "Paris is the capital of France.",
    "Berlin is the capital of Germany.",
    "The Eiffel Tower is located in Paris.",
    "Madrid is the capital of Spain.",
]
```

```python
queries = [
    {"text": "What is the capital of France?"},
    {"text": "Where is the Eiffel Tower?"},
]
```

```python


retriever = HybridRetriever()

config = create_search_config(alpha=0.5, top_k=3)

for query in queries:
    print(f"\nQUERY: {query['text']}")

    results = retriever.hybrid_search(
        query["text"],
        passages,
        alpha=config["alpha"],
        top_k=config["top_k"],
    )

    top_result = get_top_result(results)

    if top_result:
        print("\nTop Result")
        print(top_result["text"])
        print(f"Score: {top_result['score']:.4f}")

    print("\nAll Results")
    for result in results:
        print(f"Score: {result['score']:.4f}")
        print(result["text"])
        print("---")

    print("\nSparse vs Dense Contributions")
    for result in results[:3]:
        print({
            "text": result["text"],
            "sparse": result["sparse_score"],
            "dense": result["dense_score"],
            "combined": result["score"],
        })
```
